## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

### What the starter data shows

The starter dataset contains anonymized page-level observations. I will check the dataset size, how many pages have measurable search visibility, and how many pages show a downward trend.

These numbers help show whether the data has enough variation to investigate a page-review prioritization problem. They do not prove that refreshing a page will improve its performance.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Quick checks on the starter dataset

total_rows = len(df)
unique_pages = df["content_id"].nunique()

visible_pages = (df["impressions_90d"] > 0).sum()

declining_pages = (df["trend_direction"] == "down").sum()

decline_rate = (declining_pages / total_rows) * 100

print("Total rows:", total_rows)
print("Unique content IDs:", unique_pages)
print("Pages with impressions > 0:", visible_pages)
print("Pages with trend_direction = 'down':", declining_pages)
print(f"Share marked as down: {decline_rate:.2f}%")
```

## 4. Careful words: what I can and can't claim

*Write what your work will be able to say (observed, directional, decision-support) — and what it never will (causal proof, 'predicting Google').*

### What I can claim

The dataset contains observable signals related to content performance, including impressions, sessions, content age, position, CTR, and trend.

### What I cannot claim

I cannot claim that the model proves why a page declined.

I cannot claim that the model predicts Google's ranking algorithm.

The results are observational and should be treated as directional evidence. Final recommendations should be reviewed by a human.
